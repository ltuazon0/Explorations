### Visualizing Bridge Nodes, Articulation Points, & their Intersection

```python
def draw_subgraph(G, pos, title, params):
    """
    Generate a visualization of a subset of nodes & their network
    placements.
    
    Parameters
    ----------
    G      : networkx.classes.graph.Graph
             The graph where the node subset is drawn on.
    pos    : dict
             Spatial positions of all nodes & edges in the graph.
    title  : str
             Title of the visualization
    params : zip
             Object that contains 4-tuples that carry the node,
             its color (str) & size (scalar) when plotted, & the 
             node's label in the legend
    """
    
    plt.figure(figsize=(12,10))
    
    for nodelist, node_color, node_size, label in params:
        nx.draw_networkx_nodes(
            G, pos,
            nodelist=nodelist,
            node_color=node_color,
            node_size=node_size, alpha=0.75,
            label=label)

    nx.draw_networkx_edges(G, pos, edge_color='0.85', alpha=0.1)

    plt.title(title, fontsize=13)
    plt.axis('off')
    plt.legend(loc='upper left')
    plt.show()
```

```python
# Identifying bridge nodes & articulation points (cut vertices)
V_brg = set()
for u,v in list(nx.bridges(GCC)):
    V_brg.add(u)
    V_brg.add(v)        
V_cut = set(nx.articulation_points(GCC))

# Partitioning bridge nodes that are also cut vertices from leaf nodes
# (peripheral bridge nodes) & articulation points that have low 
# participation coefficients
if V_cut.issubset(V_brg):
    V_both = V_cut
else:
    V_both = V_brg.intersection(V_cut)
    
    V_brg = V_brg.difference(V_both) 
    V_cut = V_cut.difference(V_both)
    
print('Peripheral bridge nodes: ', list(V_brg))
print('\nBoth : ', list(V_both))
print('\nArticulation points: ', list(V_cut))
```

```python
def get_local_community(partition, u_ego):
    
    community_num = partition[u_ego]
    local_community = {u for u, comm in partition.items() if comm==community_num}
    
    return local_community
```

```python
def get_local_group(G, partition, u_ego, local=True):
    
    local_community = get_local_community(partition, u_ego)
    local_group = {partition[u_ego]: local_community}
    
    if local:
        neighbors = set(G.neighbors(u_ego))
        cross_communities = {partition[u]: u for u in neighbors if partition[u] != partition[u_ego]}
        
        if len(cross_communities) != 0:
            for c in list(cross_communities):
                v_ego = cross_communities[c]
                local_group.update(get_local_group(G, partition, v_ego, local=False))
                    
    return local_group
```

```python
def draw_local_group(G, pos, partition, u):
    local_group = get_local_group(G, partition, u)
    
    nodelists, node_colors, node_sizes, labels = [],[],[],[]
    
    for c in local_group:
        community = local_group[c]
        label_ = 'Local ' if c == partition[u] else ''
        
        nodelists.append(list(community))
        node_colors.append([node_color[u] for u in community])
        node_sizes.append(3)
        labels.append(label_ + f'Community {c}')
        
    nodelists.append([u])
    node_colors.append('0')
    node_sizes.append(30)
    labels.append('Articulation Point')
    
    title = f'Local Group of Articulation Point {v}'
    params = zip(nodelists, node_colors, node_sizes, labels)
    
    draw_subgraph(G, pos, title, params)
```

### Global Measures (LCC)

```python
def community_size(G, partition):
    
    comm_counts = Counter(partition.values())
    community_size = {u: comm_counts[partition[u]] for u in G.nodes()}
    
    return community_size
```

```python
def participation(G):
    
    partition = community_louvain.best_partition(G)
    
    participation_coef = {}
    for u in G.nodes():
        neighbors = list(G.neighbors(u))
        comm_neighbors = [partition[v] for v in neighbors]
        community_links = {c: comm_neighbors.count(c) for c in comm_neighbors}

        p_coef = 1
        for c in community_links:
            ratio = community_links[c]/G.degree(u)
            p_coef = p_coef - ratio**2
            
        participation_coef.update({u: p_coef})
    
    return participation_coef
```

```python
def basic_topology(G, partition):
    
    df = pd.DataFrame(index=G.nodes())
    
    df['degree'] = df.index.map(G.degree())
    df['degree_centrality'] = df.index.map(nx.degree_centrality(G))
    df['closeness_centrality'] = df.index.map(nx.closeness_centrality(G))
    df['betweenness_centrality'] = df.index.map(nx.betweenness_centrality(G))
    df['avg_neighbor_degree'] = df.index.map(nx.average_neighbor_degree(G))
    df['eccentricity'] = df.index.map(nx.eccentricity(G))
    df['clustering_coef'] = df.index.map(nx.clustering(G))
    df['community_sizes'] = df.index.map(community_size(G, partition))
    df['participation_coef'] = df.index.map(participation(G))
    
    return df
```

```python
def load_egonet(ego, root='./fb_circles/'):
    """
    Loads ego network & corresponding social circles.
    Takes in ego node id & outputs a Graph object of the ego node's
    network, a dataframe of each node's feature vector, & a dictionary
    of all the ego node's social circles. 
    """
    
    # Filepaths to ego network & features
    paths = {
        'edges'    : root + f'{ego}.edges',
        'circles'  : root + f'{ego}.circles',
        'egofeat'  : root + f'{ego}.egofeat',
        'feat'     : root + f'{ego}.feat',
        'featnames': root + f'{ego}.featnames'}
    
    # Load ego network
    G = nx.read_edgelist(paths['edges'], nodetype=int)
    
    # Load features of all nodes in ego network into dataframe
    egofeat = pd.read_csv(paths['egofeat'], sep=' ', header=None)
    egofeat.index = [ego]
    feat = pd.read_csv(paths['feat'], index_col=0, sep=' ', header=None)
    feat.columns = egofeat.columns
    feat = pd.concat([egofeat, feat], axis=0)
    
    # Load ego's social circles
    with open(paths['circles'], 'r') as f:
         circles = {f'{ego}-'+l.split()[0]: {int(u) for u in l.split()[1:]} for l in f}
    
    return G, circles, feat
```

```python
def in_circles(G, circles):
    """
    Computes social circle membership, if any, for each node in 
    a given ego network G from G's corresponding social circles,
    then annotating the graph with each node's circles before
    returning it. 
    """
    
    # Inverting mapping of social circles, with values being lists
    # of circles node u is a member of
    node_circles = {u: [] for u in G.nodes}
    for c in circles:
        for u in circles[c]:
            if u in G.nodes:
                node_circles[u].append(c)
            else:
                continue
    
    # Set circle membership as a node attribute
    nx.set_node_attributes(G, node_circles, 'circles')
    
    return G
```

```python
def structural_metrics(G):
    """
    Computes structural metrics for a given ego network G, then
    annotates the graph with these scores before returning it.
    """

    # Degree of a particular node
    degree = {u: G.degree(u) for u in G.nodes}
    
    # Prune leaves for better interpretations of centrality measures
    #leaves = {u for u in degree if degree[u]==1}
    #G.remove_nodes_from(leaves)
    
    # Centrality measures
    eigencentrality = nx.eigenvector_centrality(G)
    closeness = nx.closeness_centrality(G) # Efficiency of info flow
    betweenness = nx.betweenness_centrality(G) # Influence of flow
    
    # Clustering & participation coefficients
    clustering_coef = nx.clustering(G)
    participation_coef = participation(G)
    
    # Cut vertices
    articulation_pts = set(nx.articulation_points(G))
    articulation = {u: u in articulation_pts for u in G.nodes}
    
    # Use metrics to annotate graph as node attributes
    metrics = {
        'degree': degree,
        'eigencentrality': eigencentrality,
        'closeness': closeness,
        'betweenness': betweenness,
        'clustering_coef': clustering_coef,
        'participation_coef': participation_coef,
        'articulation': articulation
    }
    
    # Annotate metrics as node attributes
    for m in metrics:
        nx.set_node_attributes(G, metrics[m], m)
        
    return G
```

```python
def annotate(G, circles):
    
    G = structural_metrics(G)
    G = in_circles(G, circles)
    
    return G
```

```python
def local_metrics(G, circles):
    
    coverage = {C: G.subgraph(circles[C]).copy() for C in list(circles)}
    
    noncircle = {u for u in G.nodes if G.nodes[u]['circles']==[]}
    coverage.update({'noncircle': G.subgraph(noncircle).copy()})
    
    mu = lambda C,metric: np.mean(list(nx.get_node_attributes(C,metric).values()))
    
    local_covers = {
        C: {
            'avg_eigencentrality': mu(coverage[C], 'eigencentrality'),
            'avg_closeness': mu(coverage[C], 'eigencentrality'),
            'avg_betweenness': mu(coverage[C], 'eigencentrality'),
            'avg_clustering': mu(coverage[C], 'clustering_coef'),
            'avg_participation': mu(coverage[C], 'participation_coef'),
            'articulation_ratio': mu(coverage[C], 'articulation')
        }
        for C in list(coverage)
    }
    
    df_local = pd.DataFrame(local_covers).T
    
    return df_local
```

```python
def structural_divergence(G):
    
    def _structural_entropy(V):
        distr_art = [G.nodes[u]['articulation'] for u in V]
        ratio_art = sum(distr_art)/len(distr_art)
        p = [ratio_art if distr_art else 0.5, 1-ratio_art if distr_art else 0.5]
        ntrp = entropy(p, base=2)
        
        return ntrp
    
    def _k_metric(metric, k):
        u_metric = nx.get_node_attributes(G, metric)
        u_k = sorted(u_metric, key=u_metric.get, reverse=True)[:k]
        k_metric = sum(1 for u in u_k if G.nodes[u]['circles']==[])/k
        
        return k_metric

    
    k = int(len(G.nodes)/10)
    k_degree          = _k_metric('degree', k)
    k_betweenness     = _k_metric('betweenness', k)
    k_eigencentrality = _k_metric('eigencentrality', k)
    k_participation   = _k_metric('participation', k)
    
    circled = {u for u in G.nodes if G.nodes[u]['circles'] != []}
    
    ntrp_circled    = _structural_entropy(circled)
    ntrp_noncircled = _structural_entropy(set(G.nodes).difference(circled))
    
    data = {
        'k_degree'          : k_degree,
        'k_betweenness'     : k_betweenness,
        'k_eigencentrality' : k_eigencentrality,
        'k_participation'   : k_participation,
        'entropy_circled'   : ntrp_circled,
        'entropy_noncircled': ntrp_noncircled,
        'circled_count'     : len(circled),
        'noncircled_count'  : len(G.nodes)-len(circled)
    }
    
    return data
```

```python
def _featnames(u_ego):
    
    featnames_ = {}
    with open(f'./fb_circles/{u_ego}.featnames', 'r') as f:
        for l in f:
            parts = l.split(' ',1)
            featnames_[int(parts[0])] = parts[1][:-1]
            
    featnames = pd.DataFrame(
        data=featnames_.values(), 
        columns=['name'],
        index=featnames_.keys())
    
    return featnames
```

```python
def attribute_featnames(G, feat, u_ego):
    
    featnames = _featnames(u_ego)
    
    attrs = {}
    for u in G.nodes:
        feat_u=feat.T[u]==1
        feat_idx = feat_u[feat_u].index
        
        anonfeats_u = {}
        for j in feat_idx:
            featname = featnames.loc[j,'name']
            
            anonfeat = featname.split(' ')[0][:-11]
            anonfeat_name = featname.split(';')[-1]

            anonfeats_u[anonfeat] = anonfeat_name
        
        attrs[u] = anonfeats_u
        
    nx.set_node_attributes(G, attrs)
    
    return G
```

```python
def local_metrics(G, circles):
    
    coverage = {C: G.subgraph(circles[C]).copy() for C in list(circles)}
    
    noncircle = {u for u in G.nodes if G.nodes[u]['circles']==[]}
    coverage.update({'noncircle': G.subgraph(noncircle).copy()})
    
    mu = lambda C,metric: np.mean(list(nx.get_node_attributes(C,metric).values()))
    
    local_covers = {
        C: {
            'avg_eigencentrality': mu(coverage[C], 'eigencentrality'),
            'avg_closeness': mu(coverage[C], 'eigencentrality'),
            'avg_betweenness': mu(coverage[C], 'eigencentrality'),
            'avg_clustering': mu(coverage[C], 'clustering_coef'),
            'avg_participation': mu(coverage[C], 'participation_coef'),
            'articulation_ratio': mu(coverage[C], 'articulation')
        }
        for C in list(coverage)
    }
    
    df_local = pd.DataFrame(local_covers).T
    
    return df_local
```
